```python
import os
import shutil
import random

random.seed(42)

source = '/content/waste_dataset/Recyclable-NonRecyclable-Hazardous-Waste-Dataset-main'
destination = '/content/waste_split'

categories = ['Recyclable', 'Non-Recyclable']

for category in categories:
    images = os.listdir(os.path.join(source, category))
    random.shuffle(images)

    total = len(images)
    train_end = int(total * 0.8)
    val_end = int(total * 0.9)

    splits = {
        'train': images[:train_end],
        'validation': images[train_end:val_end],
        'test': images[val_end:]
    }

    for split, filenames in splits.items():
        target = os.path.join(destination, split, category)
        os.makedirs(target, exist_ok=True)

        for filename in filenames:
            shutil.copy2(
                os.path.join(source, category, filename),
                os.path.join(target, filename)
            )

print("Dataset successfully divided!")
```

```python
import tensorflow as tf

dataset_path = '/content/waste_split'

train_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path + '/train',
    image_size=(160, 160),
    batch_size=16,
    label_mode='binary'
)

validation_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path + '/validation',
    image_size=(160, 160),
    batch_size=16,
    label_mode='binary'
)

test_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path + '/test',
    image_size=(160, 160),
    batch_size=16,
    label_mode='binary'
)

print("Categories:", train_data.class_names)
```

```python
import tensorflow as tf

# 1. Load pre-trained MobileNetV2
base_model = tf.keras.applications.MobileNetV2(
    input_shape=(160, 160, 3),
    include_top=False,
    weights='imagenet'
)

# 2. Freeze pre-trained layers
base_model.trainable = False

# 3. Define input layer
inputs = tf.keras.Input(shape=(160, 160, 3))

# 4. Normalize image pixels
x = tf.keras.layers.Rescaling(
    1./127.5,
    offset=-1
)(inputs)

# 5. Extract image features
x = base_model(x, training=False)

# 6. Reduce extracted features
x = tf.keras.layers.GlobalAveragePooling2D()(x)

# 7. Add dropout to reduce overfitting
x = tf.keras.layers.Dropout(0.2)(x)

# 8. Binary classification output
outputs = tf.keras.layers.Dense(
    1,
    activation='sigmoid'
)(x)

# 9. Create final AI model
model = tf.keras.Model(inputs, outputs)

# 10. Compile the model
model.compile(
    optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# 11. Display model architecture
model.summary()

print("AI Waste Segregation Model Built Successfully!")
```

```python
history = model.fit(
    train_data,
    validation_data=validation_data,
    epochs=10
)
```

```python
test_loss, test_accuracy = model.evaluate(test_data)

print("Test Accuracy:", round(test_accuracy * 100, 2), "%")
print("Test Loss:", round(test_loss, 4))
```

```python
import os
import shutil
import random

source = '/content/final_dataset'
destination = '/content/final_split'

random.seed(42)

# Remove previous split if it exists
if os.path.exists(destination):
    shutil.rmtree(destination)

categories = ['Non-Recyclable', 'Recyclable']

for category in categories:
    images = os.listdir(os.path.join(source, category))
    random.shuffle(images)

    total = len(images)

    train_end = int(total * 0.8)
    val_end = int(total * 0.9)

    splits = {
        'train': images[:train_end],
        'validation': images[train_end:val_end],
        'test': images[val_end:]
    }

    for split, files in splits.items():
        folder = os.path.join(destination, split, category)
        os.makedirs(folder, exist_ok=True)

        for filename in files:
            shutil.copy2(
                os.path.join(source, category, filename),
                os.path.join(folder, filename)
            )

print("Dataset splitting completed!")

for split in ['train', 'validation', 'test']:
    print("\n", split.upper())

    for category in categories:
        folder = os.path.join(destination, split, category)
        print(category, ":", len(os.listdir(folder)))
```

```python
import tensorflow as tf

dataset_path = '/content/final_split'

train_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path + '/train',
    image_size=(160, 160),
    batch_size=16,
    label_mode='binary',
    shuffle=True,
    seed=42
)

validation_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path + '/validation',
    image_size=(160, 160),
    batch_size=16,
    label_mode='binary',
    shuffle=False
)

test_data = tf.keras.utils.image_dataset_from_directory(
    dataset_path + '/test',
    image_size=(160, 160),
    batch_size=16,
    label_mode='binary',
    shuffle=False
)

data_augmentation = tf.keras.Sequential([
    tf.keras.layers.RandomFlip('horizontal'),
    tf.keras.layers.RandomRotation(0.10),
    tf.keras.layers.RandomZoom(0.10),
    tf.keras.layers.RandomContrast(0.10)
])

print("Categories:", train_data.class_names)
print("Data augmentation created successfully!")
```

```python
import tensorflow as tf

base_model = tf.keras.applications.MobileNetV2(
    input_shape=(160, 160, 3),
    include_top=False,
    weights='imagenet'
)

# Freeze the pretrained layers initially
base_model.trainable = False

inputs = tf.keras.Input(shape=(160, 160, 3))

# Apply data augmentation
x = data_augmentation(inputs)

# Normalize image pixels
x = tf.keras.layers.Rescaling(
    1./127.5, offset=-1
)(x)

# Extract image features
x = base_model(x, training=False)

# Reduce feature dimensions
x = tf.keras.layers.GlobalAveragePooling2D()(x)

# Reduce overfitting
x = tf.keras.layers.Dropout(0.3)(x)

# Binary classification
outputs = tf.keras.layers.Dense(
    1, activation='sigmoid'
)(x)

model = tf.keras.Model(inputs, outputs)

model.compile(
    optimizer=tf.keras.optimizers.Adam(
        learning_rate=0.0001
    ),
    loss='binary_crossentropy',
    metrics=['accuracy']
)

model.summary()
```

```python
import tensorflow as tf

# Improve data loading performance
AUTOTUNE = tf.data.AUTOTUNE

train_data = train_data.prefetch(buffer_size=AUTOTUNE)
validation_data = validation_data.prefetch(buffer_size=AUTOTUNE)

# Training callbacks
callbacks = [
    tf.keras.callbacks.EarlyStopping(
        monitor='val_loss',
        patience=5,
        restore_best_weights=True
    ),

    tf.keras.callbacks.ReduceLROnPlateau(
        monitor='val_loss',
        factor=0.2,
        patience=2,
        min_lr=0.000001
    )
]

# Train the model
history = model.fit(
    train_data,
    validation_data=validation_data,
    epochs=20,
    callbacks=callbacks
)
```

```python
import numpy as np
import tensorflow as tf
from sklearn.metrics import confusion_matrix, classification_report

# Evaluate on test dataset
test_loss, test_accuracy = model.evaluate(test_data)

print("\nTEST ACCURACY:", round(test_accuracy * 100, 2), "%")
print("TEST LOSS:", round(test_loss, 4))

# Get actual labels
y_true = np.concatenate([
    labels.numpy().ravel() for images, labels in test_data
])

# Get model predictions
predictions = model.predict(test_data)
y_pred = (predictions.ravel() >= 0.5).astype(int)

# Confusion matrix
print("\nCONFUSION MATRIX:")
print(confusion_matrix(y_true, y_pred))

# Detailed performance report
print("\nCLASSIFICATION REPORT:")
print(classification_report(
    y_true,
    y_pred,
    target_names=['Non-Recyclable', 'Recyclable']
))
```
